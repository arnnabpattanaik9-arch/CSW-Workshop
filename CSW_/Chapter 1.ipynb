Write a Python program to accept a student's marks and display the corresponding grade using if-elif-else statements.

The program should accept marks between 0 and 100 and assign grades according to the following criteria:

*Marks 90 to 100% 0 Grade

Marks 80 to 89: A Grade

Plarks 70 to 79: B Grade

Harks 60 to 69: C Grade

Harks 50 to 59: D Grade

Marks 40 to 49: E Grade

Marks below 40: Fail

Display the appropriate grade based on the marks entered by the user.

```python
marks=int(input("Enter your marks :"))
print("You scored",marks,"marks")
if 0>marks or marks>100:
    print("Marks unjustified")
elif marks>=90 and marks<=100:
        print("You have got O Grdae")
elif 80<=marks<=89:
        print("You have got A Grade")
elif 70<=marks<=79:
        print("You have got B Grade")
elif 60<=marks<=69:
         print("You have got C Grade")
elif 50<=marks<=59:
        print("You have got D Grade")
elif 40<=marks<=49:
         print("You have got E Grade")
else:
  print("You have failed")


        
```

Write a program to print nth term of a fibonacci sequence.

```python
def fibo(n):
    curr=0
    prev=1
    while curr<=n:
        print(curr,end=" ")
        temp=curr
        curr=curr+prev
        prev=temp
n=int(input("Enter a number: "))
fibo(n)
```
