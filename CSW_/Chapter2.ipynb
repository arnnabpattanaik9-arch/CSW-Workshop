For join operation you must convert the individual elements of the list into string

Join all characters of python into a single string with * separators.

```python
q=["P","Y","T","H","O","N"]
print("*".join(q))
```

```python
enter = input("Enter your word: ")
result = []

# Iterate over the string using indices
for i in range(len(enter)):
    result.append(enter[i])
    # Add separator after every character, except the last one
    if i < len(enter) - 1:
        if i % 2 == 0:
            result.append("-")
        else:
            result.append("*")

# Join the list into a single string
print("".join(result))   
```

```python
word="PYTHON"
sep=['-','*']
parts=[]
for i,ch in enumerate(word):
    parts.append(ch)
    if i <len(word)-1:
        parts.append(sep[i%2])
print("".join(parts))
```
